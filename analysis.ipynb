# Viewer control and contextual relevance in AVOD advertising: analysis

Reproduces every number, table and figure in the paper from the raw platform export
(`data/avod_lab_data.csv`, 316 sessions). Run all cells from top to bottom; the code is in `code/`.
The SEM bootstrap (section 6) takes about five minutes.

Section numbers in the headings refer to the paper.

```python
import sys
from pathlib import Path
import pandas as pd

sys.path.insert(0, "code")
import prepare, sample, confirmatory, exploratory, additional, sem, figures

FIGURES = Path("figures")
FIGURES.mkdir(exist_ok=True)
raw = pd.read_csv("data/avod_lab_data.csv")
data = prepare.derive(raw)                    # adds the measures that exist only in the event log
df = prepare.analysed(data, verbose=True)     # exclusions as applied (Section 3.7)
```

## 1. Participants and manipulation checks (Section 4.1; cell means; Appendices C and D)

```python
sample.dropout(raw)
sample.describe(df)
sample.reliability(df)
sample.manipulation_checks(df)
```

```python
sample.cell_means(df)
```

```python
sample.participants_table(df)
```

## 2. Confirmatory tests (Sections 4.2-4.4)

```python
models, rows, decisions = confirmatory.five_tests(df)
confirmatory.simple_slopes(models["M3"])
```

## 3. Registered exploratory analyses (E1, E3, E4, E5, E7, perceived tracking)

```python
print("Placements (E5):"); exploratory.e5_placements(df)
print("Timing viewers chose (E7):"); exploratory.e7_chosen_timing(df)
print("Relevance across placements:"); exploratory.omnibus(df)
print("Control vs relevance (E4):"); exploratory.e4_control_vs_relevance(df)
```

```python
print("Persuasion knowledge and VOD use (E3):"); exploratory.e3_moderators(df)
print("Perceived fit (E1):"); exploratory.e1_perceived_fit(df)
print("Perceived tracking:"); exploratory.tracking(df)
```

## 4. Analyses not registered

```python
print("Equivalence tests:"); additional.equivalence(df)
print("Johnson-Neyman:"); additional.johnson_neyman(df)
print("Relevance at each placement:"); additional.relevance_by_placement(df)
```

```python
print("Choosing vs being assigned:"); additional.choosing_vs_assigned(df)
```

```python
print("Age and gender:"); additional.age_and_gender(df)
print("Attitudes and satisfaction:"); additional.downstream(df)
```

## 5. Robustness (Appendix E)

```python
additional.registered_rules(data)
additional.control_effect_checks(df)
additional.programme_checks(df, "data/catalogue_titles.json")
additional.creatives_and_completion(df)
```

## 6. Structural equation model (E1, E2, E6; Section 4.5)

```python
eff = sem.run(df)
```

## 7. Figures

```python
figures.placement(df, FIGURES / "fig_placement.png")
figures.simple_slopes(df, FIGURES / "fig_simple_slopes.png")
figures.sem_paths(eff, FIGURES / "fig_sem_paths.png")
```
